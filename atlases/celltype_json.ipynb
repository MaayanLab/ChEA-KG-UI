```python
import json
import pandas as pd
import requests
import json
import time 

# enrichr url
ENRICHR_URL = 'https://maayanlab.cloud/Enrichr/addList'

# grab gene set from enrichr using library and label
def fetch_geneset(label, library):
	res = requests.get(f'https://maayanlab.cloud/Enrichr/geneSetLibrary?libraryName={library}&mode=json&term={label}')
	if not res.ok:
		print(res)
		raise Exception(res.text)
	else:
		return res.json()

# get the url that lets you search a dataset
def get_url(term, lib, genes):
    g = "\n".join(genes)
    
    desc = f"{term} from {lib}"
	
    payload = {
        'list': (None, g),
        'description': (None, desc)
    }

    response = requests.post(ENRICHR_URL, files=payload)
    if not response.ok:
        print(response)
        raise Exception('Error analyzing gene list')
    
    data = json.loads(response.text)
    
    return f"https://maayanlab.cloud/Enrichr/enrich?dataset={data["shortId"]}"
```

```python
cell_types = pd.read_csv("./celltypes.txt", sep = "\t").set_index("Cell Type")
celllist = []
for index, row in cell_types.iterrows():
    term = row['Mapped Term']
    lib = row['Library']

    r = fetch_geneset(term, lib)
    
    time.sleep(2)
    if r == {}:
        url = ""
        print(f"{term} and {lib} not found in Enrichr")
    else:
        genes = r[term]
        if len(genes) > 1:
            url = get_url(term, lib, genes)
            time.sleep(2)
        else:
            url = ""
        
    celllist.append({
        'type':index,
        'tissue':row['Tissue'],
        'term':term,
        'library':lib,
        'url': url
    })

celljson = {"celltype":celllist}
```

```python
with open("../public/celltypes.json", 'w') as file:
    json.dump(celljson, file)
```
